```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('ForeignGifts_edu.csv',low_memory=False)
df.head()
```

```python
#UVA stats 
uva_only = df[df["Institution Name"]=="University of Virginia"]

#total count: 98
len(uva_only)

#average gift size: $226420
uva_mean = uva_only["Foreign Gift Amount"].mean()

#total: 22189238
uva_total = uva_only["Foreign Gift Amount"].sum()

#contract: 64 monetary gift: 34 
print(uva_only["Gift Type"].value_counts())

#top donors: 
uva_giftor_total = uva_only.groupby("Giftor Name")["Foreign Gift Amount"].sum()
uva_giftor_total.sort_values(ascending=False)
#Top 5: AstraZeneca, Intrnl Centre Diarrheal Disease, Haydom Lutheran Hospital d/b/a Hayd, Haydom Luther Hospital, Harvest Group Holding
print(uva_total)
print(uva_giftor_total.sort_values(ascending=False))
```

```python
#Comapring UVA to peer institutions (colleges in Virginia)
all_VA_rows=df[df["State"]=="VA"]
all_VA_rows["Institution Name"].unique()
#College of William & Mary, George Mason University, Virginia Commonwealth University, Virginia Polytechnic Institute & State University

va_total = all_VA_rows.groupby("Institution Name")["Foreign Gift Amount"].sum()
va_total.mean()
print(va_total.mean())
#38,392,997

#Comparing to UNC and University of Michigan
df[df["Institution Name"]=="University of North Carolina - Chapel Hill"]
df[df["Institution Name"]=="University of Michigan - Ann Arbor"]

peer_institutions = ["College of William & Mary", "George Mason University", "Virginia Commonwealth University", "Virginia Polytechnic Institute & State University", "University of North Carolina - Chapel Hill","University of Michigan - Ann Arbor" ]

def get_stats(college_name):
    place_only = df[df["Institution Name"]==college_name]
    count = len(place_only)
    place_mean = place_only["Foreign Gift Amount"].mean()
    place_total = place_only["Foreign Gift Amount"].sum()
    gift_count = place_only["Gift Type"].value_counts()
    place_giftor_total = place_only.groupby("Giftor Name")["Foreign Gift Amount"].sum()
    sorted_place_val = place_giftor_total.sort_values(ascending=False)
    top5_place = sorted_place_val.head(5)
    print(college_name)
    print(f"Number of rows: {count} //")
    print(f"Mean: {place_mean} //")
    print(f"Total: {place_total} //")
    print(f"Gift Types: {gift_count} //")
    print(f"Top 5 donors: {top5_place} //")

for college in peer_institutions:
    get_stats(college)



```
